```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_community.agent_toolkits.sql.toolkit import SQLDatabaseToolkit
from langchain_community.utilities.sql_database import SQLDatabase

# Fetch the secret securely
os.environ["GOOGLE_API_KEY"] = ''

# Initialize and use the model
llm_gemini = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")

databricks_db = SQLDatabase.from_databricks(
    catalog="job",
    schema="source",
    api_token = "",
    host="",
    warehouse_id=""
)

toolkit = SQLDatabaseToolkit(db=databricks_db, llm=llm_gemini)
toolkit.get_tools()
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model=llm_gemini,
    tools=toolkit.get_tools(),
)



example_query = "Where is the most number of jobs located?"

events = agent.stream(
    {"messages": [("user", example_query)]},
    stream_mode="values",
)
for event in events:
    event["messages"][-1].pretty_print()

```
